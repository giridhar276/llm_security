# PII Detection and Masking with Microsoft Presidio

This notebook detects personally identifiable information (PII) and replaces it with safe placeholders such as `<PERSON>`, `<EMAIL_ADDRESS>`, and `<PHONE_NUMBER>`.

## 1. Install the required packages

`presidio-analyzer` identifies sensitive information. `presidio-anonymizer` masks the identified values. Presidio uses an NLP engine such as spaCy to recognize entities. Run this cell once.

## 2. Import and initialize Presidio

The analyzer detects PII and returns its type, position, and confidence score. The anonymizer replaces the detected values.

```python
from presidio_analyzer import AnalyzerEngine
from presidio_anonymizer import AnonymizerEngine

analyzer = AnalyzerEngine()
anonymizer = AnonymizerEngine()
```

## 3. Provide sample text

The example contains several common PII types: a name, email address, phone number, IP address, and credit-card number. Use only synthetic data during demonstrations.

```python
text = """
My name is John Smith.
My email address is john.smith@example.com.
My phone number is +91 98765 43210.
My IP address is 192.168.1.25.
My credit card number is 4111 1111 1111 1111.
"""

print(text)
```

## 4. Detect PII

Presidio examines the text and returns one result for every detected entity. The confidence score ranges from 0 to 1; a higher value indicates greater confidence.

```python
results = analyzer.analyze(
    text=text,
    language="en"
)

for result in results:
    detected_value = text[result.start:result.end]
    print(
        f"Type: {result.entity_type:<20} "
        f"Value: {detected_value:<30} "
        f"Score: {result.score:.2f}"
    )
```

## 5. Mask the detected PII

The anonymizer replaces every detected value with a label representing its entity type. The original text remains unchanged in the `text` variable.

```python
masked_result = anonymizer.anonymize(
    text=text,
    analyzer_results=results
)

print(masked_result.text)
```

## 6. Compare the original and masked text

This final comparison makes it easy to verify that sensitive information was removed before the text is stored, logged, or sent to an LLM.

```python
print("ORIGINAL TEXT")
print("-" * 60)
print(text)

print("MASKED TEXT")
print("-" * 60)
print(masked_result.text)
```
